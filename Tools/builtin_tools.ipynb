## Custom Tool


```python
from langchain_community.tools import tool

def multiply(a,b):
    """Multiply 2 numbers"""
    return a*b
```

```python
def multiply(a:int,b:int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
@tool
def multiply(a:int,b:int) -> int:
    """Multiply two numbers"""
    return a*b
```

```python
result=multiply.invoke({'a':2,'b':3})
result
```

```python
print(multiply.description)
print(multiply.args)
print(multiply.name)
```

```python
print(multiply.args_schema.model_json_schema())
```

## Using Structured Tool

```python
from langchain_community.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class Multiply(BaseModel):
    a: int = Field(json_schema_extra={"required": True, "description": "The first number to add."})  
    b: int = Field(json_schema_extra={"required": True, "description": "The second number to add."})  
```

```python
def multiply(a:int,b:int)->int:
    return a*b
    
```

```python
multiply_tool=StructuredTool.from_function(
    func=multiply,
    name="multiply",
    description="Multiply 2 numbers",
    args_schema=Multiply
)
```

```python
multiply_tool.invoke({'a':3,'b':7})
```

```python
print(multiply_tool.description)
print(multiply_tool.args)
print(multiply_tool.name)
print(multiply_tool.args_schema.model_json_schema())
```

## Toolkit

```python
from langchain_core.tools import tool
```

```python
@tool
def multiply(a:int,b:int)->int:
    """Multply 2 numbers"""
    return a*b

@tool
def add(a:int,b:int)->int:
    """Add 2 numbers"""
    return a+b
```

```python
class MathToolKit:
    def get_tools(self):
        return[add,multiply]
```

```python
toolkit=MathToolKit()
tools=toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.invoke({'a':3,"b":4}))
```
