```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
import requests
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.tools import DuckDuckGoSearchRun
search_tool=DuckDuckGoSearchRun()
results=search_tool.invoke('what is the main news for today?')
```

```python
@tool
def getConversionFactor(target_currency:str)-> float:
    """This function fetches the currency conversion factor of a given base currency and target currency"""
    url=f"https://v6.exchangerate-api.com/v6/{os.getenv("API_KEY")}/latest/{target_currency}"
    response=requests.get(url)
    return response.json()
```

```python
llm=ChatGoogleGenerativeAI(
    model='gemini-3.5-flash-lite',
    google_api_key=os.getenv("GEMINI_API_KEY")
)
```

```python
llm.invoke('hi')
```

```python
from langchain.agents import create_agent
from langsmith import Client

# Pull a prompt template
client=Client()
prompt = client.pull_prompt("hwchase17/react",dangerously_pull_public_prompt=True)
```

```python
prompt
```

```python
agent=create_agent(
    model=llm,
    tools=[search_tool,getConversionFactor],
    system_prompt=prompt.template,
    debug=True
)
```

```python
response=agent.invoke({"messages":{'role':'user',"content":"Find the currency code of Nepal and then find the value of 1 USD in NPR"}})
```

```python
response['messages']
```
